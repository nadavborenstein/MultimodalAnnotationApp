```python
from PIL import Image
from glob import glob
import pandas as pd
```

```python
images_path = "/Users/knf792/gits/MMFC/data/tweet_images/"
notes_path = "/Users/knf792/gits/AnnotationWebsite/data/multimodal_tweets_balanced.csv"
save_path = "static/resized_images/"
max_image_height = 600
```

```python
notes = pd.read_csv(notes_path)
images = notes["image_name"].tolist()
print(f"Found {len(images)} images in the csv file.")
```

```python
for image in images:
    try:
        image_path = images_path + image
        img = Image.open(image_path).convert("RGB")
        width, height = img.size
        if height > max_image_height:
            new_height = max_image_height
            new_width = int((new_height / height) * width)
            img = img.resize((new_width, new_height))
            print(f"Resized {image} to {new_width}x{new_height}")
        else:
            print(f"No resizing needed for {image}")
        img.save(save_path + image)
    except Exception as e:
        print(f"Error processing {image}: {e}")
```
